```python
# Update Portfolio and Asset References

Notebook to inspect the existing image and PDF files in the `TrilogiMaknaCreative` folder, then map them to the `index.html` sections:
- Galeri Portofolio
- Creative Assets & Compro
- Paket Layanan / Daftar Harga

This notebook helps verify and correct file references based on actual files present in the workspace.
```

```python
from pathlib import Path

root = Path(r'C:\Users\Ikhwan Farhat\Documents\TrilogiMaknaCreative')
print(f'Project root: {root}')
print('Exists:', root.exists())

# Utility to list files in a folder with optional extension filtering

def list_files(folder, extensions=None):
    path = root / folder
    if not path.exists():
        return []
    return sorted([f.name for f in path.iterdir() if f.is_file() and (extensions is None or f.suffix.lower() in extensions)])

print('\nTop-level files:')
for file in sorted([f.name for f in root.iterdir() if f.is_file()]):
    print('-', file)

print('\nSubfolders:')
for folder in sorted([f.name for f in root.iterdir() if f.is_dir()]):
    print('-', folder)
```

```python
# 2. Scan TrilogiMaknaCreative Folder

portfolio_folders = ['graduate', 'konser', 'SPORT', 'ULANG TAHUN', 'WEDDING', 'GATHERING']
print('Portfolio folders and files:')
for folder in portfolio_folders:
    files = list_files(folder, extensions={'.jpg', '.jpeg', '.png'})
    print(f'\n{folder}:')
    for f in files:
        print('-', f)

pdf_files = [f.name for f in root.iterdir() if f.is_file() and f.suffix.lower() == '.pdf']
print('\nPDF files:')
for f in sorted(pdf_files):
    print('-', f)
```

```python
# 3. Load and Display Galeri Portofolio Images

from IPython.display import display, Image, HTML

# Use a small sample for visual validation
sample_images = {
    'Graduate': list_files('graduate', extensions={'.jpg', '.jpeg', '.png'})[:6],
    'Konser': list_files('konser', extensions={'.jpg', '.jpeg', '.png'})[:4],
    'Sport': list_files('SPORT', extensions={'.jpg', '.jpeg', '.png'})[:4],
    'Birthday': list_files('ULANG TAHUN', extensions={'.jpg', '.jpeg', '.png'})[:4],
    'Wedding': list_files('WEDDING', extensions={'.jpg', '.jpeg', '.png'})[:4],
    'Gathering': list_files('GATHERING', extensions={'.jpg', '.jpeg', '.png'})[:4],
}

for category, files in sample_images.items():
    if not files:
        print(f'No images found for {category}')
        continue
    print(f'\n{category} ({len(files)} sample images):')
    gallery_html = '<div style="display:flex;flex-wrap:wrap;gap:10px;">'
    for filename in files:
        img_path = root / category.lower() if category != 'Birthday' else root / 'ULANG TAHUN'
        if category == 'Sport':
            img_path = root / 'SPORT'
        img_path = img_path / filename
        gallery_html += f'<div style="width:150px;"><img src="{img_path.as_posix()}" style="width:100%;height:auto;object-fit:cover;border-radius:8px;"><p style="color:#eee;font-size:12px;">{filename}</p></div>'
    gallery_html += '</div>'
    display(HTML(gallery_html))
```

```python
# 4. Load and Embed Creative Assets and Compro PDFs

pdf_assets = [
    'compro TrilogiMaknaCreative.pdf',
    'Katalog TM.Creative.pdf',
    'Moodboard.pdf',
    'PORTO TRILOGI MAKNA CREATIVE.pdf.pdf',
    'PRICELIST GRADUATION 2026 .pdf.pdf',
    'Pricelist TrilogiMaknaCreative.pdf',
    'pricelistWEDDING.pdf'
]

print('Available PDFs:')
for pdf in pdf_assets:
    exists = (root / pdf).exists()
    print(f'- {pdf}:', 'FOUND' if exists else 'MISSING')

# Build HTML for quick PDF link preview
pdf_html = '<div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(250px,1fr));gap:12px;">'
for pdf in pdf_assets:
    file_path = root / pdf
    status = 'FOUND' if file_path.exists() else 'MISSING'
    pdf_html += f'<div style="padding:12px;border:1px solid #444;background:#111;border-radius:12px;"><h4 style="color:#fff;font-size:14px;">{pdf}</h4><p style="color:#aaa;font-size:12px;">{status}</p>'
    if file_path.exists():
        pdf_html += f'<a href="{file_path.as_posix()}" target="_blank" style="color:#36E2F5;text-decoration:none;">Open PDF</a>'
    pdf_html += '</div>'
pdf_html += '</div>'
display(HTML(pdf_html))
```
